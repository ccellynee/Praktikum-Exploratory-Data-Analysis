## **E. Proyek Akhir: EDA Starter Project**

### **1. Data Loading & Inspection**

```python
import pandas as pd  # mengimpor library Pandas

df = pd.read_csv('data_penjualan_atk.csv')  # membaca dataset dari file CSV

print(df.head())  # menampilkan 5 baris pertama
df.info()  # menampilkan informasi data dan tipe data
print(df.describe())  # menampilkan statistik data angka
print(df.shape)  # menampilkan jumlah baris dan kolom
```

Dari pemeriksaan awal, dataset memiliki 20 baris dan 4 kolom. Terdapat data kosong pada kolom produk dan terjual.

### **2. Data Cleaning**

```python
python
print(df.isnull().sum())  # menghitung jumlah data kosong pada setiap kolom

df['terjual'] = df['terjual'].fillna(0)  # mengganti data terjual yang kosong menjadi 0

df = df.dropna(subset=['produk'])  # menghapus baris yang memiliki produk kosong

print(df.duplicated().sum())  # menghitung jumlah data duplikat

df = df.drop_duplicates()  # menghapus data yang benar-benar sama

df['harga'] = df['harga'].astype(int)  # memastikan harga bertipe integer

print(df.dtypes)  # menampilkan tipe data setiap kolom
```

**Alasan Pembersihan:**
* terjual yang kosong diisi *0* karena tidak ada jumlah penjualan yang tercatat pada data tersebut.
* produk yang kosong dihapus karena nama produk diperlukan untuk mengidentifikasi dan mengelompokkan data.
* Data duplikat dihapus agar data tidak dihitung lebih dari satu kali.
* Tipe data harga dipastikan integer agar dapat digunakan dalam perhitungan.

Setelah proses cleaning, dataset menjadi *18 baris*.

### **3. Data Manipulation**

```python
#a. Filtering

python
laris = df[df['terjual'] > 5]  # mengambil produk yang terjual lebih dari 5

print(laris)  # menampilkan hasil filtering
```

Hasilnya terdapat *10 data* dengan jumlah penjualan lebih dari 5.


```python
# b. Sorting

python
urut = df.sort_values(by='terjual', ascending=False)  # mengurutkan dari penjualan terbesar

print(urut)  # menampilkan data yang sudah diurutkan
```

Produk dengan jumlah penjualan tertinggi adalah *Pensil, yaitu **10 buah*.

```python
# c. Kolom Turunan

python
df['total_pendapatan'] = df['harga'] * df['terjual']  # menghitung total pendapatan

print(df)  # menampilkan data dengan kolom pendapatan
```


Kolom total_pendapatan digunakan untuk mengetahui pendapatan dari setiap produk.

```python
# d. Groupby / Agregasi

python
ringkasan = df.groupby('produk')['total_pendapatan'].sum()  # menjumlahkan pendapatan berdasarkan produk

print(ringkasan)  # menampilkan total pendapatan setiap produk
```

Berdasarkan hasil groupby, *Buku Gambar* memiliki total pendapatan tertinggi, yaitu *Rp104.000*.
