## **LATIHAN 1**

```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

### **ANALISIS** - latihan 1

NumPy array dapat melakukan operasi matematika pada seluruh elemen secara langsung. Sedangkan list Python biasa tidak dapat dikalikan dengan bilangan desimal seperti `0.9` dan akan menghasilkan error.

Hal tersebut menunjukkan bahwa NumPy lebih praktis untuk mengolah data numerik karena operasi dapat dilakukan pada banyak data sekaligus.


## **LATIHAN 2**

```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)

print(df)
```

### **Analisis** - latihan 2

Terdapat data kosong pada kolom `menu` dan `terjual`. Jika langsung digunakan, data tersebut dapat membuat hasil analisis menjadi kurang akurat. Maka dari itu, data kosong perlu ditangani terlebih dahulu agar dataset lebih lengkap dan hasil analisis lebih dapat dipercaya.


## **LATIHAN 3**

```python
import pandas as pd  # mengimpor library Pandas

df = pd.read_csv('data_kantin_20_baris.csv')  # membaca data dari file CSV

print(df.head())  # menampilkan 5 baris pertama
print(df.info())  # menampilkan informasi data dan jumlah data yang terisi
print(df.describe())  # menampilkan statistik dari data angka
print(df.shape)  # menampilkan jumlah baris dan kolom
```

### **Analisis** - latihan 3

Berdasarkan hasil df.info(), kolom menu dan terjual memiliki jumlah data yang kurang dari total 20 baris. Hal ini menunjukkan bahwa terdapat data yang kosong pada kedua kolom tersebut.

## **LATIHAN 4**

```python
print(df.isnull().sum())  # menghitung jumlah data kosong pada setiap kolom

df['terjual'] = df['terjual'].fillna(0)  # mengganti data terjual yang kosong menjadi 0

df = df.dropna(subset=['menu'])  # menghapus baris yang memiliki data menu kosong

print(df)  # menampilkan data setelah dibersihkan
```

### **Analisis** - latihan 4

Kolom terjual menggunakan fillna(0) karena data yang kosong masih dapat dianggap sebagai tidak ada penjualan yang tercatat. Sedangkan data dengan menu yang kosong dihapus karena nama menu diperlukan untuk mengetahui jenis makanan atau minuman yang dianalisis.


## **LATIHAN 5**

```python
print(df.duplicated().sum())  # menghitung jumlah data yang duplikat

df = df.drop_duplicates()  # menghapus data yang benar-benar sama

df['harga'] = df['harga'].astype(int)  # mengubah tipe data harga menjadi integer

print(df.dtypes)  # menampilkan tipe data setiap kolom
```

### **Analisis** - latihan 5

Setelah menggunakan `drop_duplicates()`, jumlah data tetap 19 baris karena tidak terdapat data yang benar-benar sama. Tipe data perlu diperiksa agar perhitungan dapat dilakukan dengan benar, misalnya saat menghitung total pendapatan dari harga dikali jumlah barang terjual.


## **LATIHAN 6**

```python
laris = df[df['terjual'] > 20]  # mengambil menu yang terjual lebih dari 20

urut = df.sort_values(by='terjual', ascending=False)  # mengurutkan data dari penjualan terbesar

df['total_pendapatan'] = df['harga'] * df['terjual']  # menghitung total pendapatan

ringkasan = df.groupby('menu')['total_pendapatan'].sum()  # menjumlahkan pendapatan berdasarkan menu

print(ringkasan)  # menampilkan ringkasan total pendapatan setiap menu
```

### **Analisis** - latihan 6

Menu dengan total pendapatan tertinggi adalah Ayam Geprek, yaitu Rp900.000. Hasil ini dapat menjadi pertimbangan kantin dalam mengatur persediaan dan mengevaluasi menu yang memiliki penjualan lebih rendah.
